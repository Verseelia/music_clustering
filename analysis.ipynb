````python
X = df[audio_features]
````

````python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)

X_scaled = pd.DataFrame(
    X_scaled,
    columns=audio_features
)

````

````python
X_scaled.describe().round(2)
````

````python
from sklearn.cluster import KMeans

inertia = []

k_values = range(2,11)

for k in k_values:

    model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=20
    )

    model.fit(X_scaled)

    inertia.append(model.inertia_)
````

````python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))

plt.plot(
    k_values,
    inertia,
    marker="o",
    linewidth=2
)

plt.xlabel("Number of Clusters (K)")
plt.ylabel("Inertia")

plt.title("Elbow Method")

plt.grid(True)

plt.show()
````

## Elbow Method

The **Elbow Method** is used to determine the optimal number of clusters (`K`) for K-Means clustering.

### Observation

From the plot:

- The inertia decreases significantly from **K = 2 to K = 4**.
- After **K = 4**, the decrease in inertia becomes more gradual.
- The curve starts to flatten around **K = 4**.

### Interpretation

The point where the rate of decrease in inertia changes significantly is considered the **elbow point**.

In this case, the elbow appears to be around:

> **K = 4**

This suggests that **4 clusters** may be a suitable choice for the K-Means model.

### Conclusion

Based on the Elbow Method, we will consider **K = 3,4** as a potential optimal number of clusters.

However, the Elbow Method alone may not provide a definitive answer. Therefore, we will compare different values of `K` using additional clustering evaluation metrics such as:



````python
from sklearn.model_selection import train_test_split

X_sample, _ = train_test_split(
    X_scaled,
    train_size=30000,
    random_state=42
)
````

````python
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

silhouette_scores = []

for k in k_values:

    model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    labels = model.fit_predict(X_sample)

    score = silhouette_score(
        X_sample,
        labels
    )

    silhouette_scores.append(score)

    print(f"K = {k}, Silhouette Score = {score:.4f}")
````

## Silhouette Score Analysis

The **Silhouette Score** is used to evaluate the quality of clustering. It measures how well each data point fits within its assigned cluster compared with other clusters.

- A score closer to **+1** indicates well-separated and compact clusters.
- A score around **0** indicates overlapping clusters.
- A negative score indicates that data points may have been assigned to the wrong clusters.

### Silhouette Scores for Different K Values

| Number of Clusters (K) | Silhouette Score |
|-------------------------|------------------|
| 2 | 0.2013 |
| 3 | **0.2398** |
| 4 | 0.2290 |
| 5 | 0.1861 |
| 6 | 0.1881 |
| 7 | 0.1642 |
| 8 | 0.1681 |
| 9 | 0.1681 |
| 10 | 0.1753 |

### Observation

- The highest Silhouette Score is obtained at **K = 3**, with a score of **0.2398**.
- **K = 4** has the second-highest score of **0.2290**.
- After **K = 4**, the Silhouette Score generally decreases.
- The scores remain positive for all tested values of `K`, indicating that the clusters have some meaningful separation.

### Interpretation

Although the **Elbow Method suggested K = 4**, the Silhouette Score suggests that **K = 3 provides better-defined and more separated clusters**.

The difference between K = 3 and K = 4 is relatively small:

- K = 3 → **0.2398**
- K = 4 → **0.2290**

Therefore, both values are reasonable candidates, but **K = 3 is preferred based on the Silhouette Score**.

### Conclusion

> **Based on the Silhouette Score, K = 3 is the optimal choice among the tested values because it achieves the highest score of 0.2398.**

````python
from sklearn.cluster import DBSCAN
from sklearn.metrics import silhouette_score
import pandas as pd

eps_values = [1.2, 1.3, 1.4, 1.5, 2.0]

results = []

for eps in eps_values:

    dbscan = DBSCAN(
        eps=eps,
        min_samples=10
    )

    labels = dbscan.fit_predict(X_dbscan)

    # Remove noise (-1)
    mask = labels != -1

    X_clustered = X_dbscan[mask]
    labels_clustered = labels[mask]

    # Silhouette requires at least 2 clusters
    if len(set(labels_clustered)) > 1:
        silhouette = silhouette_score(
            X_clustered,
            labels_clustered
        )
    else:
        silhouette = None

    n_clusters = len(set(labels)) - (1 if -1 in labels else 0)

    noise_count = sum(labels == -1)
    noise_pct = noise_count / len(labels) * 100

    results.append({
        "eps": eps,
        "clusters": n_clusters,
        "noise_count": noise_count,
        "noise_pct": noise_pct,
        "silhouette_score": silhouette
    })

results_df = pd.DataFrame(results)

results_df
````

## DBSCAN Parameter Selection and Final Model

DBSCAN (Density-Based Spatial Clustering of Applications with Noise) was selected to identify groups of similar songs based on their audio features. Unlike K-Means, DBSCAN does not require the number of clusters to be specified in advance and can also identify observations that do not belong to any dense cluster as noise.

### Why Parameter Tuning Was Required

DBSCAN has two important parameters:

- **`eps`**: Defines the maximum distance between two points for them to be considered neighbors. A larger `eps` creates larger neighborhoods and can cause multiple groups to merge into a single cluster, while a smaller `eps` creates stricter neighborhoods and can result in more observations being classified as noise.

- **`min_samples`**: Specifies the minimum number of neighboring observations required for a point to be considered part of a dense region. Increasing `min_samples` makes DBSCAN more restrictive and generally results in more points being classified as noise.

Since the appropriate values of these parameters are data-dependent, they were tuned systematically rather than selecting them arbitrarily.

---

### Step 1: Tuning `eps`

Initially, `min_samples` was fixed at **10**, and different values of `eps` were evaluated using a 5,000-row sample of the dataset.

The following values were tested:

- `eps = 1.2`
- `eps = 1.3`
- `eps = 1.4`
- `eps = 1.5`
- `eps = 2.0`

The results were:

|-----|-------------------:|------------:|--------:|-----------------:|
| 1.2 | 5 | 1517 | 30.34% | 0.194 |
| 1.3 | 6 | 1131 | 22.62% | 0.223 |
| **1.4** | **5** | **833** | **16.66%** | **0.278** |
| 1.5 | 5 | 628 | 12.56% | 0.234 |
| 2.0 | 1 | 125 | 2.50% | Not applicable |

### Interpretation of `eps` Tuning

When `eps` was reduced from 2.0 to smaller values, DBSCAN became more restrictive.

At `eps = 2.0`, almost all observations were assigned to a single cluster (97.5%), indicating that the neighborhood radius was too large to identify meaningful separation between groups.

At the other extreme, `eps = 1.2` classified approximately 30% of the observations as noise, indicating that the neighborhood radius was becoming too restrictive.

Among the tested values, **`eps = 1.4` produced the highest Silhouette Score of 0.278** while maintaining a moderate noise proportion of 16.66%. Therefore, `eps = 1.4` was selected for the next stage of parameter tuning.

> `eps = 1.1` produces a more diverse cluster structure than `eps = 1.2`, with a dominant cluster of **68.93%**, a secondary cluster of **8.97%**, and another cluster containing **2.07%** of the observations.

Although this provides better separation of density regions, the **19.64% noise rate** remains relatively high.

````python
from sklearn.cluster import DBSCAN
from sklearn.metrics import silhouette_score
import pandas as pd

eps_values = [1.0, 1.1, 1.2]

results = []

for eps in eps_values:

    dbscan = DBSCAN(
        eps=eps,
        min_samples=15
    )

    labels = dbscan.fit_predict(X_dbscan_30k)

    # Remove noise
    mask = labels != -1

    X_clustered = X_dbscan_30k[mask]
    labels_clustered = labels[mask]

    # Calculate silhouette score
    if len(set(labels_clustered)) > 1:
        silhouette = silhouette_score(
            X_clustered,
            labels_clustered
        )
    else:
        silhouette = None

    n_clusters = len(set(labels)) - (
        1 if -1 in labels else 0
    )

    noise_count = (labels == -1).sum()
    noise_percentage = noise_count / len(labels) * 100

    results.append({
        "eps": eps,
        "clusters": n_clusters,
        "noise_count": noise_count,
        "noise_pct": noise_percentage,
        "silhouette_score": silhouette
    })

results_df = pd.DataFrame(results)

results_df
````
